Набор данных, содержащий информацию о 1000 химических соединениях и их эффективности против вируса гриппа, измеренной с помощью параметров IC50, CC50 и SI.

1. **IC50 (полумаксимальная ингибирующая концентрация)**: это концентрация вещества, необходимая для ингибирования определенного биологического или биохимического процесса на 50%. Чем ниже значение IC50, тем более эффективным считается соединение в ингибировании целевого процесса.

2. **CC50 (полумаксимальная цитотоксическая концентрация)**: это концентрация вещества, при которой 50% клеток погибают или теряют свою функцию. Это значение используется для оценки токсичности соединения. Более высокое значение CC50 указывает на меньшую токсичность.

3. **SI (индекс селективности)**: это отношение CC50 к IC50 (SI = CC50 / IC50). Индекс селективности показывает, насколько избирательно соединение действует на вирусные клетки по сравнению с нормальными клетками. Более высокое значение SI указывает на более высокую селективность и потенциально более безопасное соединение для использования.

эти термины проще:

1. **IC50**: Представьте, что у вас есть вещество, которое может замедлить или остановить действие вируса. IC50 — это количество этого вещества, которое нужно, чтобы замедлить действие вируса наполовину. Меньшее значение IC50 означает, что нужно меньше вещества, чтобы достичь этого эффекта, и это говорит о том, что вещество более эффективно против вируса.

2. **CC50**: Теперь представьте, что это же вещество может быть вредным для обычных клеток. CC50 — это количество вещества, которое нужно, чтобы повредить или убить половину нормальных клеток. Более высокое значение CC50 означает, что вещество менее токсично для обычных клеток.

Таким образом, если вещество имеет низкое значение IC50 и высокое значение CC50, это значит, что оно эффективно против вируса и в то же время безопасно для нормальных клеток.

1. **MaxAbsEStateIndex, MaxEStateIndex, MinAbsEStateIndex, MinEStateIndex**: Эти параметры связаны с электронным состоянием атомов в молекуле. Они описывают максимальные и минимальные значения индексов электронного состояния, которые могут быть абсолютными или относительными.

2. **qed (Quantitative Estimate of Drug-likeness)**: Это количественная оценка "лекарственности" соединения, основанная на различных химических свойствах.

3. **SPS (Synthetic Accessibility Score)**: Оценка синтетической доступности соединения, то есть насколько легко или сложно его синтезировать.

4. **MolWt, HeavyAtomMolWt, ExactMolWt**: Молекулярный вес соединения. HeavyAtomMolWt учитывает только тяжелые атомы (не водород), а ExactMolWt — точный молекулярный вес.

5. **NumValenceElectrons, NumRadicalElectrons**: Количество валентных электронов и радикальных электронов в молекуле.

6. **MaxPartialCharge, MinPartialCharge, MaxAbsPartialCharge, MinAbsPartialCharge**: Эти параметры описывают максимальные и минимальные частичные заряды атомов в молекуле.

7. **FpDensityMorgan1, FpDensityMorgan2, FpDensityMorgan3**: Плотность отпечатков (fingerprints) Morgan на различных радиусах. Эти параметры используются для описания структуры молекулы.

8. **BCUT2D_MWHI, BCUT2D_MWLOW, BCUT2D_CHGHI, BCUT2D_CHGLO, BCUT2D_LOGPHI, BCUT2D_LOGPLOW, BCUT2D_MRHI, BCUT2D_MRLOW**: Это параметры, связанные с молекулярными дескрипторами BCUT, которые оценивают различные аспекты молекулярной структуры, такие как молекулярный вес, заряд, логарифм распределения и молекулярный рефрактивный индекс.

9. **AvgIpc, BalabanJ, BertzCT**: Различные индексы, которые оценивают сложность и связность молекулы.

10. **Chi0, Chi0n, Chi0v, Chi1, Chi1n, Chi1v, Chi2n, Chi2v, Chi3n, Chi3v, Chi4n, Chi4v**: Индексы Хи (Chi indices), которые описывают различные аспекты топологии молекулы.

11. **HallKierAlpha**: Параметр, связанный с альфа-индексом Холла-Киера, который оценивает гибкость молекулы.

12. **Ipc**: Индекс информационной сложности молекулы.

13. **Kappa1, Kappa2, Kappa3**: Индексы формы молекулы, которые оценивают ее трехмерную структуру.

14. **LabuteASA**: Площадь доступной поверхности молекулы, рассчитанная по методу Лабута.

15. **PEOE_VSA1, PEOE_VSA2, ..., PEOE_VSA14**: Параметры, связанные с объемной поверхностью молекулы, рассчитанные с использованием метода PEOE (Partial Equalization of Orbital Electronegativities).

16. **SMR_VSA1, SMR_VSA2, ..., SMR_VSA10**: Параметры, связанные с объемной поверхностью молекулы, рассчитанные с использованием метода SMR (Simple Molecular Refractivity).

17. **SlogP_VSA1, SlogP_VSA2, ..., SlogP_VSA12**: Параметры, связанные с объемной поверхностью молекулы, рассчитанные с использованием метода SlogP (логарифм коэффициента распределения).

18. **TPSA**: Полярная поверхность молекулы, которая может влиять на ее способность проникать через клеточные мембраны.

19. **EState_VSA1, EState_VSA2, ..., EState_VSA11**: Параметры, связанные с объемной поверхностью молекулы, рассчитанные с использованием метода EState (Electronic State).

20. **VSA_EState1, VSA_EState2, ..., VSA_EState10**: Параметры, связанные с объемной поверхностью молекулы, рассчитанные с использованием метода VSA (Volume Surface Area) и EState.

21. **FractionCSP3**: Доля атомов углерода в молекуле, имеющих sp3-гибридизацию.

22. **HeavyAtomCount**: Количество тяжелых атомов (не водород) в молекуле.

23. **NHOHCount, NOCount**: Количество функциональных групп, содержащих N-H или O-H связи, и количество атомов азота и кислорода.

24. **NumAliphaticCarbocycles, NumAliphaticHeterocycles, NumAliphaticRings**: Количество алифатических карбоциклов, гетероциклов и колец.

25. **NumAromaticCarbocycles, NumAromaticHeterocycles, NumAromaticRings**: Количество ароматических карбоциклов, гетероциклов и колец.

26. **NumHAcceptors, NumHDonors**: Количество акцепторов и доноров водородных связей.

27. **NumHeteroatoms**: Количество гетероатомов в молекуле.

28. **NumRotatableBonds**: Количество вращаемых связей в молекуле.

29. **NumSaturatedCarbocycles, NumSaturatedHeterocycles, NumSaturatedRings**: Количество насыщенных карбоциклов, гетероциклов и колец.

30. **RingCount**: Общее количество колец в молекуле.

31. **MolLogP**: Логарифм коэффициента распределения (гидрофобность) молекулы.

32. **MolMR**: Молекулярный рефрактивный индекс.

33. **fr_Al_COO, fr_Al_OH, ..., fr_urea**: Эти параметры описывают наличие различных функциональных групп в молекуле, таких как карбоксильные группы, гидроксильные группы, амины, эфиры и другие.

Эти дескрипторы используются для анализа и моделирования химических свойств соединений, а также для предсказания их биологической активности.

```python
def get_missing_stat(df): 
    missing_count = df.isna().sum()
    missing_percentage = (missing_count / len(df)) * 100
    missing_data = pd.DataFrame({
        'Missing Count': missing_count,
        'Missing Percentage': round(missing_percentage, 2)
    })
    return missing_data[missing_data['Missing Count'] > 0]

get_missing_stat(df)
```

Пропусков в данных практически нет. Заполним их средним значением:

```python
def print_corr_columns(df):
    correlation_matrix = df.corr()
    ic50_corr = correlation_matrix['IC50, mM']
    cc50_corr = correlation_matrix['CC50, mM']
    si_corr = correlation_matrix['SI']
    high_corr_pairs = []
    for i in range(len(correlation_matrix)):
        for j in range(i+1, len(correlation_matrix)):
            if abs(correlation_matrix.iloc[i, j]) > 0.99:
                high_corr_pairs.append((correlation_matrix.columns[i], correlation_matrix.columns[j], correlation_matrix.iloc[i, j]))

    for pair in high_corr_pairs:
        print(f"Признаки '{pair[0]}' и '{pair[1]}' коррелируют с коэффициентом {pair[2]:.2f}")
        print(f"IC50 '{pair[0]}': {ic50_corr[pair[0]]}, '{pair[1]}': {ic50_corr[pair[1]]}")
        print(f"CC50 '{pair[0]}': {cc50_corr[pair[0]]}, '{pair[1]}': {cc50_corr[pair[1]]}")
        print(f"SI '{pair[0]}': {si_corr[pair[0]]}, '{pair[1]}': {si_corr[pair[1]]}")
        print("\n")

print_corr_columns(df)
```

Много признаков коррелирует между собой с коэффициентом привыщающий 0.99.

```python
import sys
import os

sys.path.append(os.path.abspath('lib'))

from preprocessing_pipeline import create_preprocessing_pipeline

from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.feature_selection import RFE
```

```python
preprocessing_pipeline = create_preprocessing_pipeline()
preprocessing_pipeline
```

```python
X = df.drop(columns=['IC50, mM', 'CC50, mM', 'SI'])
y1 = df['IC50, mM']
y2 = df['CC50, mM']
y3 = df['SI']

X_transformed = preprocessing_pipeline.fit_transform(X)

X_train_ic, X_test_ic, y_train_ic, y_test_ic = train_test_split(X_transformed, y1, test_size=0.2, random_state=42)
X_train_cc, X_test_cc, y_train_cc, y_test_cc = train_test_split(X_transformed, y2, test_size=0.2, random_state=42)
X_train_si, X_test_si, y_train_si, y_test_si = train_test_split(X_transformed, y3, test_size=0.2, random_state=42)
```

```python
def selectKBest(model, X_train, X_test, y_train, y_test):
    selected_features_list = []
    mae_scores = []
    r2_scores = []
    
    for n_features in range(5, (X_train.shape[1] + 1) - 10):
        selector = SelectKBest(score_func=f_regression, k=n_features)
        X_train_selected = selector.fit_transform(X_train, y_train)
        X_test_selected = selector.transform(X_test)
    
        model.fit(X_train_selected, y_train)
        y_pred = model.predict(X_test_selected)

        selected_features = X_train.columns[selector.get_support()]
        mae = mean_absolute_error(y_test, y_pred)
        r2 = r2_score(y_test, y_pred)
        
        selected_features_list.append(selected_features)
        mae_scores.append(mae)
        r2_scores.append(r2)
    return (mae_scores, r2_scores, selected_features_list)    
```

```python
def selectRFE(model, X_train, X_test, y_train, y_test):
    selected_features_list = []
    mae_scores = []
    r2_scores = []
    
    for n_features in range(5, (X_train.shape[1] + 1) - 10):
        rfe = RFE(estimator=model, n_features_to_select=n_features)
        X_train_selected = rfe.fit_transform(X_train, y_train)
        X_test_selected = rfe.transform(X_test)
        
        model.fit(X_train_selected, y_train)
        
        y_pred = model.predict(X_test_selected)

        selected_features = X_train.columns[rfe.support_]
        mae = mean_absolute_error(y_test, y_pred)
        r2 = r2_score(y_test, y_pred)

        selected_features_list.append(selected_features)
        mae_scores.append(mae)
        r2_scores.append(r2)
    return (mae_scores, r2_scores, selected_features_list)    
```

```python
def drawSelection(scores, name, target_name):
    # Построение графика
    plt.figure(figsize=(14, 8))
    plt.plot(range(1, len(scores) + 1), scores, marker='o')
    plt.title(f'Зависимость {name} {target_name} от количества выбранных признаков')
    plt.xlabel('Количество выбранных признаков')
    plt.ylabel(name)
    plt.xticks(range(10, len(scores) + 1, 10))
    for x in range(10, len(scores) + 1, 10):
        plt.axvline(x=x, color='gray', linestyle='--', linewidth=0.7)
    plt.show()
```
